```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 设置中文字体和数学公式渲染
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['mathtext.fontset'] = 'stix'  # 数学公式用 STIX 字体，支持希腊字母和上下标
```

```python
# 1. 椭球面: x²/a² + y²/b² + z²/c² = 1
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, np.pi, 50)
a, b, c = 3, 2, 1.5
x = a * np.outer(np.cos(u), np.sin(v))
y = b * np.outer(np.sin(u), np.sin(v))
z = c * np.outer(np.ones_like(u), np.cos(v))

ax.plot_surface(x, y, z, alpha=0.7, cmap='viridis')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title(rf'椭球面: $x^2/{a}^2 + y^2/{b}^2 + z^2/{c}^2 = 1$')
plt.tight_layout()
plt.show()
```

```python
# 2. 单叶双曲面: x²/a² + y²/b² - z²/c² = 1
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(-2, 2, 50)
a, b, c = 2, 2, 3
x = a * np.cosh(v) * np.outer(np.cos(u), np.ones_like(v))
y = b * np.cosh(v) * np.outer(np.sin(u), np.ones_like(v))
z = c * np.outer(np.ones_like(u), np.sinh(v))

ax.plot_surface(x, y, z, alpha=0.7, cmap='coolwarm')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title(rf'单叶双曲面: $x^2/{a}^2 + y^2/{b}^2 - z^2/{c}^2 = 1$')
plt.tight_layout()
plt.show()
```

```python
# 3. 双叶双曲面: x²/a² + y²/b² - z²/c² = -1
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 100)
v1 = np.linspace(0.3, 3, 30)
v2 = np.linspace(-3, -0.3, 30)
a, b, c = 2, 2, 3

for v, color in [(v1, 'blue'), (v2, 'red')]:
    x = a * np.sinh(v) * np.outer(np.cos(u), np.ones_like(v))
    y = b * np.sinh(v) * np.outer(np.sin(u), np.ones_like(v))
    z = c * np.outer(np.ones_like(u), np.cosh(v))
    ax.plot_surface(x, y, z, alpha=0.6, color=color)

ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title(rf'双叶双曲面: $x^2/{a}^2 + y^2/{b}^2 - z^2/{c}^2 = -1$')
plt.tight_layout()
plt.show()
```

```python
# 4. 椭圆抛物面: z = x²/a² + y²/b²
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 100)
r = np.linspace(0, 3, 50)
a, b = 2, 1.5
x = a * r * np.outer(np.cos(u), np.ones_like(r))
y = b * r * np.outer(np.sin(u), np.ones_like(r))
z = r**2 * np.outer(np.ones_like(u), np.ones_like(r))

ax.plot_surface(x, y, z, alpha=0.7, cmap='plasma')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title(rf'椭圆抛物面: $z = x^2/{a}^2 + y^2/{b}^2$')
plt.tight_layout()
plt.show()
```

```python
# 5. 双曲抛物面(马鞍面): z = x²/a² - y²/b²
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
a, b = 2, 2
Z = X**2/a**2 - Y**2/b**2

ax.plot_surface(X, Y, Z, alpha=0.7, cmap='RdYlBu')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title(rf'双曲抛物面(马鞍面): $z = x^2/{a}^2 - y^2/{b}^2$')
plt.tight_layout()
plt.show()
```

```python
# 6. 螺旋线
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

t = np.linspace(0, 6*np.pi, 500)
R, h = 2, 1
x = R * np.cos(t)
y = R * np.sin(t)
z = h * t

ax.plot(x, y, z, 'b-', linewidth=2)
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title(r'螺旋线: $(R\cos t,\ R\sin t,\ ht)$')
plt.tight_layout()
plt.show()
```

```python
# 7. Möbius 带
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(-0.5, 0.5, 20)
U, V = np.meshgrid(u, v)
R = 3
X = (R + V*np.cos(U/2)) * np.cos(U)
Y = (R + V*np.cos(U/2)) * np.sin(U)
Z = V * np.sin(U/2)

ax.plot_surface(X, Y, Z, alpha=0.8, cmap='rainbow')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.set_title('Möbius 带')
plt.tight_layout()
plt.show()
```
